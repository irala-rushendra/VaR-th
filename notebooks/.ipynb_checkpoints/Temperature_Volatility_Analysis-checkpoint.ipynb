```python
import numpy as np
import pandas as pd
from pathlib import Path
from datetime import datetime, timezone
import glob
import requests
import matplotlib.pyplot as plt
from pandas.plotting import register_matplotlib_converters
import seaborn as sns
from statsmodels.tsa.seasonal import STL, seasonal_decompose
from statsmodels.tsa.stattools import adfuller, kpss

register_matplotlib_converters()
sns.set_style("darkgrid")
plt.style.use('seaborn-v0_8-whitegrid')
```

```python
url = "https://raw.githubusercontent.com/datasets/global-temp/main/data/monthly.csv"
test_horizon = 120  
root = Path.cwd().parent
raw_save_dest = root / "data" / "raw"
raw_save_dest.mkdir(parents=True, exist_ok=True)

proc_save_dest = root / "data" / "processed"
proc_save_dest.mkdir(parents=True, exist_ok=True)
```

```python
def processed_gistemp(raw: pd.DataFrame, src: str = "GISTEMP") -> pd.Series:
    """
    Takes raw long format df with columns
    Returns a processed pd.Series indexed by monthly datetime
    """
    proc_data = raw[raw['Source'] == src].copy()
    proc_data['Year'] = pd.to_datetime(proc_data['Year'], format="%Y-%m")
    proc_data = proc_data.sort_values(by=['Year'], ascending=True).set_index("Year")
    proc_data.index.freq = "MS"

    anomaly_series = proc_data["Mean"].rename("anomaly")
    return anomaly_series

def check_missing_months(series: pd.Series) -> pd.DateTimeIndex:
    expected_range = pd.date_range(series.index.min(), series.index.max(), freq="MS")
    missing_dates = expected_range.difference(series.index)
    return missing_dates

def train_test_split_chrono(series: pd.Series, test_months: int = 120):
    train_chunk = series[:-test_months]
    test_chunk = series[-test_months:]
    return train_chunk, test_chunk

def save_processed(series: pd.Series, path: str = proc_save_dest/"gistemp_monthly.csv"):
    proc_data = series.to_csv(path, index=True)

def get_raw_data(url: str=None, save: bool = True) -> pd.DataFrame:
    """
    Downloading dataset from url and caching it to data/raw
    """
    try:
        data = pd.read_csv(url)
    except Exception as e:
        print(f"Error parsing the DataFrame: {e}")
        return pd.DataFrame()
    if not save:
        return data
    else:
        timestamp = datetime.now().isoformat().replace(":","-")
        fname = f"global_temp_monthly_{timestamp}.csv"
        data.to_csv(raw_save_dest/fname, index=False)
        return data
 
def load_data() -> pd.DataFrame:
    """
    Load Cached data
    """
    files = sorted(glob.glob(raw_save_dest/"global_temp_monthly_*.csv"))
    if not files:
        print("No cached data found\n")
        return pd.DataFrame()
    df = pd.read_csv(files[-1])
    return df
```

```python
raw = get_raw_data(url, True)
print(raw)
series = processed_gistemp(raw)
print(series)
missing = check_missing_months(series)
print(missing)
print(f"Missing: {len(missing)}")   

save_processed(series)
train, test = train_test_split_chrono(series)
print(len(train), len(test))        
series.describe()
```

```python
def evaluate_stationarity(ts, name):
    adf_stat, adf_p, _, _, adf_crit, _ = adfuller(ts.dropna())
    print(f"[ADF Test]  P-value: {adf_p:.6f}")
    print(f"            Result:  {'STATIONARY' if adf_p < 0.05 else 'NON-STATIONARY'} (Alpha = 0.05)")
    
    # KPSS Test
    kpss_stat, kpss_p, _, kpss_crit = kpss(ts.dropna(), regression='c', nlags='auto')
    print(f"[KPSS Test] P-value: {kpss_p:.6f}")
    print(f"            Result:  {'STATIONARY' if kpss_p > 0.05 else 'NON-STATIONARY'} (Alpha = 0.05)")

evaluate_stationarity(series, "Raw GISTEMP Series")
evaluate_stationarity(series.diff(), "First-Differenced Series (d=1)")
evaluate_stationarity(stl_res.resid, "STL Residual Series (Volatility Input)")
```

```python
import scipy.stats as stats
import statsmodels.api as sm
import numpy as np
import matplotlib.pyplot as plt

train_residuals = fitted_model.resid

lb_test = sm.stats.acorr_ljungbox(train_residuals, lags=[12], return_df=True)
lb_pvalue = lb_test['lb_pvalue'].values[0]
print(f"[Ljung-Box Test] P-value: {lb_pvalue:.6f}")
print(f"                 Result:  {'NO AUTOCORRELATION' if lb_pvalue > 0.05 else 'REMAINING DEPENDENCY'} (Alpha=0.05)")

jb_stat, jb_pvalue = stats.jarque_bera(train_residuals)
print(f"[Jarque-Bera]    P-value: {jb_pvalue:.6f}")
print(f"                 Result:  {'NORMALLY DISTRIBUTED' if jb_pvalue > 0.05 else 'FAT-TAILED / NON-NORMAL'} (Alpha=0.05)")

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sm.qqplot(train_residuals, line='s', ax=axes[0])
axes[0].set_title('Normal Q-Q Plot')
axes[0].grid(True, alpha=0.3)

axes[1].hist(train_residuals, bins=30, density=True, alpha=0.6, color='purple', edgecolor='black')
xmin, xmax = axes[1].get_xlim()
x_axis = np.linspace(xmin, xmax, 100)
axes[1].plot(x_axis, stats.norm.pdf(x_axis, train_residuals.mean(), train_residuals.std()), color='black', lw=2, linestyle='--')
axes[1].set_title('Residual Density vs Normal Curve')
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
from arch import arch_model

train_res_input = train_residuals.dropna()

print("      FITTING GARCH-FAMILY MODELS")

# 1. Fit Standard GARCH(1,1)
garch_mod = arch_model(train_res_input, p=1, q=1, vol='GARCH', dist='normal')
garch_fit = garch_mod.fit(disp='off')
print(f"Standard GARCH(1,1) -> AIC: {garch_fit.aic:.4f} | BIC: {garch_fit.bic:.4f}")

# 2. Fit GJR-GARCH(1,1) (Set o=1 to enable asymmetry parameter)
gjr_mod = arch_model(train_res_input, p=1, o=1, q=1, vol='GARCH', dist='normal')
gjr_fit = gjr_mod.fit(disp='off')
print(f"GJR-GARCH(1,1)      -> AIC: {gjr_fit.aic:.4f} | BIC: {gjr_fit.bic:.4f}")

# 3. Fit EGARCH(1,1)
egarch_mod = arch_model(train_res_input, p=1, q=1, vol='EGARCH', dist='normal')
egarch_fit = egarch_mod.fit(disp='off')
print(f"EGARCH(1,1)          -> AIC: {egarch_fit.aic:.4f} | BIC: {egarch_fit.bic:.4f}")

# 4. Determine the Winner
models = {
    'Standard GARCH(1,1)': (garch_fit.aic, garch_fit),
    'GJR-GARCH(1,1)': (gjr_fit.aic, gjr_fit),
    'EGARCH(1,1)': (egarch_fit.aic, egarch_fit)
}
best_model_name = min(models, key=lambda k: models[k][0])
print(f" Best AIC : {best_model_name}")
```

```python
import pandas as pd
import numpy as np
from arch import arch_model
from sklearn.metrics import mean_squared_error

# 1. Prepare full data sequence (scaled by 100 for optimizer stability)
scaled_full_res = full_sample_residuals * 100
train_len = len(train_residuals.dropna())
test_len = len(scaled_full_res) - train_len

history = list(scaled_full_res.iloc[:train_len])
predictions = []
naive_benchmark = []

# Global training baseline variance to serve as our Naive Benchmark
constant_variance_baseline = np.var(history)

print("Starting Walk-Forward Volatility Forecast Evaluation...")
# 2. Execute Walk-Forward Loop over the 10-year Holdout Window
for t in range(test_len):
    # Fit model on current rolling historical window
    model = arch_model(history, p=1, q=1, vol='EGARCH', dist='normal')
    model_fit = model.fit(disp='off')
    
    # Forecast the immediate next step volatility value
    pred_vol = model_fit.forecast(horizon=1).variance.iloc[-1, 0]
    predictions.append(pred_vol)
    naive_benchmark.append(constant_variance_baseline)
    
    # Pull the true observed next-step shock value from the test set and append to history
    true_shock = scaled_full_res.iloc[train_len + t]
    history.append(true_shock)

# 3. Calculate Proxy for True Variance Shocks in the Test Set (Squared Residuals)
true_test_variance = (scaled_full_res.iloc[train_len:].values) ** 2

# Scale all variances back to original unit coordinates for accurate report tracking
predictions_unscaled = np.array(predictions) / 10000
naive_unscaled = np.array(naive_benchmark) / 10000
true_test_variance_unscaled = true_test_variance / 10000

# 4. Score Model Predictability Performance vs Benchmark
egarch_mse = mean_squared_error(true_test_variance_unscaled, predictions_unscaled)
naive_mse = mean_squared_error(true_test_variance_unscaled, naive_unscaled)

print("  OUT-OF-SAMPLE VOLATILITY SKILL SCORE")
print(f"EGARCH(1,1) Forecast Variance MSE:    {egarch_mse:.7f}")
print(f"Naive Constant-Variance Baseline MSE: {naive_mse:.7f}")
if egarch_mse < naive_mse:
    skill_pct = ((naive_mse - egarch_mse) / naive_mse) * 100
    print(f"Result: EGARCH beats baseline by {skill_pct:.2f}% predictive edge.")
else:
    print("Result: Naive baseline wins. Global mean aggregation lacks out-of-sample predictability.")
```

```python

import pandas as pd
import io
import requests

def fetch_oni_raw():
    """Fetches the official NOAA CPC Oceanic Niño Index dataset bypassing the 403 security wall."""
    # This is the exact static data path for the ascii matrix on NOAA CPC
    oni_url = "https://www.cpc.ncep.noaa.gov/data/indices/oni.ascii.txt"
    
    # Establish standard browser identity headers to completely clear the 403 Forbidden trap
    headers = {
        'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36'
    }
    
    print(f"Streaming live ONI data safely from NOAA CPC...")
    response = requests.get(oni_url, headers=headers)
    response.raise_for_status() # Force an error if connection drops
    
    # Wrap the string response into an in-memory text buffer so pandas can parse the fixed widths
    data_buffer = io.StringIO(response.text)
    
    # Parse fixed-width whitespace delimited columns
    df_raw = pd.read_csv(data_buffer, sep=r'\s+', engine='python')
    return df_raw


def clean_oni(df_raw):
    """Maps overlapping 3-month seasons to standard calendar months."""
    # Map NOAA 3-letter seasons to corresponding central calendar months
    season_to_month = {
        'DJF': 1,  'JFM': 2,  'FMA': 3,  'MAM': 4,
        'AMJ': 5,  'MJJ': 6,  'JJA': 7,  'JAS': 8,
        'ASO': 9,  'SON': 10, 'OND': 11, 'NDJ': 12
    }
    
    df_clean = df_raw.copy()
    # Strip any whitespace from season strings
    df_clean['SEAS'] = df_clean['SEAS'].str.strip()
    
    # Map text seasons to integer months
    df_clean['Month'] = df_clean['SEAS'].map(season_to_month)
    
    # Create valid datetime index (Setting day to 1st of the month)
    df_clean['Date'] = pd.to_datetime(df_clean['YR'].astype(str) + '-' + df_clean['Month'].astype(str) + '-01')
    
    # Isolate the anomaly column (ANOM represents the ONI score)
    df_oni = df_clean.set_index('Date')[['ANOM']].rename(columns={'ANOM': 'ONI'})
    
    # Validation step: Verify against a known extreme El Niño event (Winter 1997-1998 peak)
    peak_1997 = df_oni.loc['1997-11-01':'1998-01-01']
    print(peak_1997)
    
    return df_oni

# 1. Execute the ENSO Pipeline
df_raw_oni = fetch_oni_raw()
df_oni_clean = clean_oni(df_raw_oni)

# 2. Merge GISTEMP + ONI (Assuming your processed GISTEMP dataframe is named 'df_gistemp')
# This step automatically aligns dates and constraints the usable range to 1950+
df_gistemp = series.copy()
# 2. Convert your Series into a DataFrame on-the-fly to enable table joining
# 2. Re-enforce proper DatetimeIndex properties during table compilation
df_merged = pd.DataFrame(series).rename(columns={series.name: 'Mean'}).join(df_oni_clean, how='inner')
df_merged.index = pd.to_datetime(df_merged.index)

print("     MERGED DATASET OVERVIEW (1950+)")
print(f"Data availability window: {df_merged.index.min().strftime('%Y-%m')} to {df_merged.index.max().strftime('%Y-%m')}")
print("\nFirst 5 Rows:")
print(df_merged.head())


```

```python
import pandas as pd
import numpy as np
from arch import arch_model
import statsmodels.api as sm
from sklearn.metrics import mean_squared_error

# 1. Prepare data structures (scaled by 100 for optimizer stability)
scaled_full_res = df_merged['Scaled_Residuals']
train_len = len(df_merged) - test_horizon

history = list(scaled_full_res.iloc[:train_len])
garch_x_predictions = []

print("Running Two-Step Walk-Forward Loop for Volatility Analysis...")
# 2. Walk-Forward Loop over the 10-year Test Block
for t in range(test_horizon):
    # Slice historical training window dynamically
    current_train_res = scaled_full_res.iloc[:train_len + t]
    current_train_oni_sq = df_merged['ONI_sq'].iloc[:train_len + t]
    
    # Step A: Fit standard EGARCH(1,1) on current training slice
    egarch_mod = arch_model(current_train_res, p=1, q=1, vol='EGARCH', dist='normal')
    egarch_fit = egarch_mod.fit(disp='off')
    
    # Extract the historical conditional variance series generated by EGARCH
    h_t = egarch_fit.conditional_volatility ** 2
    
    # Step B: Fit an OLS model combining EGARCH variance + ONI_sq to predict squared shocks
    X_ols = pd.DataFrame({
        'GARCH_Var': h_t,
        'ONI_sq': current_train_oni_sq
    })
    X_ols = sm.add_constant(X_ols) # Add intercept term
    y_ols = current_train_res ** 2
    
    ols_fit = sm.OLS(y_ols, X_ols).fit()
    
    # Step C: Forecast 1-Step Ahead Variance
    # Forecast the standard baseline EGARCH variance for upcoming month
    pred_egarch_var = egarch_fit.forecast(horizon=1).variance.iloc[-1, 0]
    upcoming_oni_sq = df_merged['ONI_sq'].iloc[train_len + t]
    
    # Combine baseline forecast with upcoming ONI shock using our OLS coefficients
    next_features = [1.0, pred_egarch_var, upcoming_oni_sq]
    pred_combined_var = np.dot(ols_fit.params, next_features)
    
    # Ensure variance never falls below zero due to parameter estimation shifts
    garch_x_predictions.append(max(0.0001, pred_combined_var))

# 3. Scale test variance proxies and predictions back down to original unit dimensions
true_test_var_unscaled = (scaled_full_res.iloc[train_len:].values) ** 2 / 10000
garch_x_predictions_unscaled = np.array(garch_x_predictions) / 10000

# 4. Generate Performance Results Scorecard
garch_x_mse = mean_squared_error(true_test_var_unscaled, garch_x_predictions_unscaled)

print("     GARCH-X OUT-OF-SAMPLE SKILL SCORE")
print(f"Two-Step EGARCH-X Variance MSE: {garch_x_mse:.7f}")
print(f"Compare this value to your plain EGARCH MSE ({egarch_mse:.7f}).")
```

```python
import torch
import torch.nn as nn
import numpy as np
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error
import matplotlib.pyplot as plt

torch.manual_seed(42)
np.random.seed(42)

raw_values = series.values.reshape(-1, 1)
train_size = len(series) - 120  # Lock an identical 120-month (10-year) holdout window

train_data = raw_values[:train_size]
test_data = raw_values[train_size:]

scaler = MinMaxScaler(feature_range=(-1, 1))
train_scaled = scaler.fit_transform(train_data)
test_scaled = scaler.transform(test_data) # Project test bounds through historical bounds

# Construct sliding window frames (12-month sequence lookback memory)
def create_sequences(data, lookback=12):
    X, y = [], []
    for i in range(len(data) - lookback):
        X.append(data[i:(i + lookback)])
        y.append(data[i + lookback])
    return torch.FloatTensor(np.array(X)), torch.FloatTensor(np.array(y))

lookback_window = 12
X_train, y_train = create_sequences(train_scaled, lookback_window)
print(f"LSTM Training Tensor shape: {X_train.shape} -> [Sequences, Lookback Months, Features]")


class ClimateLSTM(nn.Module):
    def __init__(self, input_dim=1, hidden_dim=32, num_layers=2, output_dim=1):
        super(ClimateLSTM, self).__init__()
        # 2-layer architecture with 32 hidden units to balance depth without overfitting
        self.lstm = nn.LSTM(input_dim, hidden_dim, num_layers, batch_first=True)
        self.linear = nn.Linear(hidden_dim, output_dim)
        
    def forward(self, x):
        lstm_out, _ = self.lstm(x)
        # Pass the final time lag's hidden layer state directly into your linear prediction mapping
        predictions = self.linear(lstm_out[:, -1, :])
        return predictions

model_lstm = ClimateLSTM()
criterion = nn.MSELoss()
optimizer = torch.optim.Adam(model_lstm.parameters(), lr=0.01)

# Fit model parameters
model_lstm.train()
for epoch in range(120):
    optimizer.zero_grad()
    outputs = model_lstm(X_train)
    loss = criterion(outputs, y_train)
    loss.backward()
    optimizer.step()
    
    if (epoch + 1) % 30 == 0:
        print(f"Epoch [{epoch+1}/120] -> Internal Network Convergence Loss: {loss.item():.5f}")

model_lstm.eval()
# Collect the trailing historical training window to ignite forecasting vectors
inputs = list(train_scaled[-lookback_window:].flatten())
lstm_predictions = []

# Roll the neural network forward recursively over your 10-year future window
for _ in range(120):
    current_window = torch.FloatTensor(np.array(inputs[-lookback_window:]).reshape(1, lookback_window, 1))
    with torch.no_grad():
        pred_scaled = model_lstm(current_window).item()
        lstm_predictions.append(pred_scaled)
        inputs.append(pred_scaled) # Feed prediction back to serve as next time-lag feature

# Invert scaling vectors back to true temperature anomaly values
lstm_final_forecast = scaler.inverse_transform(np.array(lstm_predictions).reshape(-1, 1)).flatten()

# Score model accuracy against actual test points
actual_test_values = series.iloc[train_size:].values
lstm_mse = mean_squared_error(actual_test_values, lstm_final_forecast)

print(f"\nLSTM Deep Learning Mean Forecast Test MSE: {lstm_mse:.5f}")
```
